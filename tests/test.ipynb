### Testing *args and **kwargs of the Company class

```python
company = cp.Company(name="PharmaCorp")
company.summarize_activity("Researching new drugs", "Launching a public health campaign")
```

For the previous bit of code to work, we wrote the class method something like this:

```python
### DO NOT ATTEMPT TO RUN

class Company:

    ...

    def summarize_activity(self, *args, **kwargs):
        """
        Summarizes company activities and additional information.

        Parameters:
        - *args: A list of activities related to the company.
        - **kwargs: Additional information, like location or date.
        """
        print(f"\nActivity Summary for {self.name}:")

        if args:
            print("Activities:")
            for activity in args:
                print(f" - {activity}")

        if kwargs:
            print("Additional Information:")
            for key, value in kwargs.items():
                print(f" - {key.capitalize()}: {value}")


###
```

- **Positional Arguments:** *args is used to pass a variable number of positional arguments to the constructor.

- **Keyword Arguments:** **kwargs is used to pass a variable number of keyword arguments to the constructor.

```python
company.summarize_activity(
    "Researching new drugs", "Launching a public health campaign",  # these are positional arguments
    location="New York", date="2024-10-27"  # these are keyword arguments
)
```
